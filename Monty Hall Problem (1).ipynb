```python
import numpy as np
def monty_hall_game(num_games, door_pick, keep_switch): 
    wins = 0
    door_set = [1,2,3]
    for n in range(0, num_games):
        open_door_set = [1,2,3] # doors that can be opened
        unchosen_door_set = [1,2,3] # doors that remain unchosen
        unchosen_door_set.remove(door_pick) 
        win_door = np.random.choice(door_set, 1) # define winning door
        # Defining doors that will be opened for zonks
        if door_pick == win_door: 
            open_door_set.remove(win_door)
        else:
            open_door_set.remove(win_door) 
            open_door_set.remove(door_pick)
        open_door = np.random.choice(open_door_set, 1) # Open door for zonk 
        unchosen_door_set.remove(open_door) 
        if keep_switch == "k": 
            if door_pick == win_door: 
                wins += 1 
        if keep_switch == "s": 
            if unchosen_door_set[0] == win_door: 
                wins += 1
    print(float(wins)/float(num_games)*100,"%")

monty_hall_game(10**5,1,"k")
```
